Get the tools

```python
import altair as alt
import pandas as pd
alt.renderers.enable("mimetype")
```

Grabbing the data and cleaning the columns, then displaying table

```python
winedataR = pd.read_csv(
    r"C:\Users\NathanKrueger\Downloads\winequality-red.csv",
    sep=";"
)

winedataW = pd.read_csv(
    r"C:\Users\NathanKrueger\Downloads\winequality-white.csv",
    sep=";"
)

winedataR.columns = winedataR.columns.str.strip()
winedataW.columns = winedataW.columns.str.strip()

from IPython.display import HTML, display

display(HTML("<h3>Red Wine</h3>"))
display(winedataR.head(10))

display(HTML("<h3>White Wine</h3>"))
display(winedataW.head(10))
```

First vis, alcohol % compared to density

```python
chart1 = alt.Chart(winedataR).mark_point().encode(
    x = alt.X("alcohol", scale = alt.Scale(domain=[7, 15])), # this is the percentage of alcohol
    y = alt.Y("density", scale = alt.Scale(domain=[.98, 1.040])),
    color = alt.Color("quality:N").scale(scheme = "plasma")
).properties(
    title = "Red Wine: Alcohol Content Compared to its Density",
     width = 500,
    height = 500
)

chart2 = alt.Chart(winedataW).mark_point().encode(
    x = alt.X("alcohol", scale = alt.Scale(domain=[7, 15])), # this is the percentage of alcohol
    y = alt.Y("density", scale = alt.Scale(domain=[.98, 1.040])),
    color = alt.Color("quality:N").scale(scheme = "plasma")
).properties(
    title = "White Wine: Alcohol Content Compared to its Density",
    width = 500,
    height = 500
)

alt.hconcat(chart1, chart2)
```

Vis 2, finding the sweet spot via %

```python
alt.data_transformers.disable_max_rows()

winedataR['type'] = 'red'
winedataW['type'] = 'white'

paired = pd.concat([winedataR, winedataW])

heatmap = alt.Chart(paired).mark_rect().encode(
    alt.X('alcohol:Q', bin=alt.Bin(maxbins=20), title='Alcohol (%)'),
    alt.Y('quality:O', title='Quality'),
    alt.Color('count()', scale=alt.Scale(scheme='purpleblue'), title='Count')
).properties(
    width=300,
    height=300
).facet(
    column=alt.Column('type:N', title='Wine Type')
).properties(
    title='Alcohol vs Quality: Red vs White Wines'
)

heatmap
```

Vis 3, noting chlorides

```python
alt.data_transformers.disable_max_rows()

winedataR['type'] = 'red'
winedataW['type'] = 'white'

paired = pd.concat([winedataR, winedataW]).reset_index(drop=True)
chart = alt.Chart(paired).mark_boxplot(size=50).encode(
    x=alt.X('chlorides:Q', title='chlorides'),
    y=alt.Y('type:N', title='Wine Type'),
    color=alt.Color('type:N', scale=alt.Scale(
        domain=['red', 'white'],
        range=['darkred', 'gold']
    ))
).properties(
    width=700,
    height=200
)

chart
```

Final visualization, comparing sulfur dioxide density

```python
red_filtered = winedataR[(winedataR['total sulfur dioxide'] <= 300) & 
                         (winedataR['free sulfur dioxide'] <= 100)]

white_filtered = winedataW[(winedataW['total sulfur dioxide'] <= 300) & 
                           (winedataW['free sulfur dioxide'] <= 100)]

red_chart = alt.Chart(red_filtered).mark_rect().encode(
    x=alt.X('total sulfur dioxide:Q', bin=alt.Bin(maxbins=30), 
            scale=alt.Scale(domain=[0, 300]), title='Total Sulfur Dioxide'),
    y=alt.Y('free sulfur dioxide:Q', bin=alt.Bin(maxbins=30), 
            scale=alt.Scale(domain=[0, 100]), title='Free Sulfur Dioxide'),
    color=alt.Color('count():Q', scale=alt.Scale(scheme='viridis'), title='Density')
).properties(
    title='Red Wine: Free vs Total Sulfur Dioxide Density'
)

white_chart = alt.Chart(white_filtered).mark_rect().encode(
    x=alt.X('total sulfur dioxide:Q', bin=alt.Bin(maxbins=30), 
            scale=alt.Scale(domain=[0, 300]), title='Total Sulfur Dioxide'),
    y=alt.Y('free sulfur dioxide:Q', bin=alt.Bin(maxbins=30), 
            scale=alt.Scale(domain=[0, 100]), title='Free Sulfur Dioxide'),
    color=alt.Color('count():Q', scale=alt.Scale(scheme='blues'), title='Density')
).properties(
    title='White Wine: Free vs Total Sulfur Dioxide Density'
)

red_chart & white_chart


combined_chart4 = alt.hconcat(red_chart, white_chart)
combined_chart4.save("cc4.png")

from IPython.display import Image, display

display(Image(filename="cc4.png"))
```
